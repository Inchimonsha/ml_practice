## Парсинг простой страницы с цитатами

Задача: собрать текст и автора цитат с сайта http://quotes.toscrape.com/ (или аналогичного).

Основные действия: отправить запрос, получить HTML, найти нужные элементы (теги с цитатами и авторами), извлечь текст.

Цель: освоить функции поиска тегов (find, find_all), навигацию по DOM, получение атрибутов и текста.



```python
import requests
from bs4 import BeautifulSoup

url = 'http://quotes.toscrape.com/'
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')

quotes = soup.find_all('div', class_='quote')
for q in quotes:
    text = q.find('span', class_='text').get_text()
    author = q.find('small', class_='author').get_text()
    print(f'"{text}" — {author}')
```

## Сбор данных с сайтов с товарами

Задача: парсить страницу интернет-магазина, получить список товаров с названием, ценой, ссылкой на изображение.

Учиться работать с классами и ID в HTML, извлекать атрибуты (например, href, src).

Можно сохранить данные в CSV или базу для дальнейшего анализа.


```python
import pandas as pd
import requests
from bs4 import BeautifulSoup

url = "https://www.muztorg.ru/promo/bestsellers/"

response = requests.get(url)
html = response.text

soup = BeautifulSoup(html, 'lxml')
products = soup.find_all("article", class_="catalog-card")

for product in products:
    prod_info_attrs = product.find("div", class_="catalog-card__info")
    product_info = [prod_info_attrs.find("a", class_="catalog-card__name").text, prod_info_attrs.find("a", class_="catalog-card__category").text]
    product_price = product.find("div", "catalog-card__price-block").text
    product_img_link = product.find("img", "catalog-card__picture").get("data-src")
    print(*product_info, product_price, product_img_link)
```
