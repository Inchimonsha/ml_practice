``````python
import pandas as pd
``````

``````python
file_path = 'data/feed-views.log'
views = pd.read_csv(file_path, sep='\t', names=['datetime', 'user'])

views['datetime'] = pd.to_datetime(views['datetime'])

year = views['datetime'].dt.year
month = views['datetime'].dt.month
day = views['datetime'].dt.day
hour = views['datetime'].dt.hour
minute = views['datetime'].dt.minute
second = views['datetime'].dt.second
``````

``````python
def get_daytime(hour):
    if (hour >= 0) & (hour <= 3.59):
        return 'night'
    elif (hour >= 4.) & (hour <= 6.59):
        return 'early morning'
    elif (hour >= 7.) & (hour <= 10.59):
        return 'morning'
    elif (hour >= 11.) & (hour <= 16.59):
        return 'afternoon'
    elif (hour >= 17.) & (hour <= 19.59):
        return 'early evening'
    elif (hour >= 20.) & (hour <= 23.59):
        return 'evening'
    else:
        return 'unknown'

views['daytime'] = views['datetime'].dt.hour.apply(get_daytime)

print(views.count())

print(views.value_counts('daytime'))
``````

``````python
views.sort_values(by='datetime', key=lambda x: x.dt.hour * 60 * 60 + x.dt.minute * 60 + x.dt.second, ascending=True, inplace=True)
``````

``````python
import pandas as pd
import gc
``````

``````python
data_file_path = './data/fines.csv'
df = pd.read_csv(data_file_path)
df
``````

``````python
%%timeit
def calc_loop_fry(df):
    res = []
    for i in range(len(df)):
        fn = df.iloc[i]['Fines']
        rf = df.iloc[i]['Refund']
        year = df.iloc[i]['Year']
        val = fn / rf * year if rf != 0 else None
        res.append(val)
    return res

df['loop_res_fry'] = calc_loop_fry(df)
gc.collect()
``````

``````python
%%timeit
def calc_iterrows_fry(df):
    res = []
    for i, item in df.iterrows():
        fn = item['Fines']
        rf = item['Refund']
        year = item['Year']
        val = fn / rf * year if rf != 0 else None
        res.append(val)
    return res

df['iterrows_res_fry'] = calc_iterrows_fry(df)
gc.collect()
``````

``````python
%%timeit
df['apply_res_fry'] = df.apply(lambda item: item['Fines'] / item['Refund'] * item['Year'] if item['Refund'] != 0 else None, axis=1)
gc.collect()
``````

``````python
%%timeit
df['series_res_fry'] = df['Fines'] / df['Refund'] * df['Year']
gc.collect()
``````

``````python
%%timeit
fn = df['Fines'].values
rf = df['Refund'].values
yr = df['Year'].values
df['values_res_fry'] = fn / rf * yr
gc.collect()
``````
