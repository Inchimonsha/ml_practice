```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("ekibee/car-sales-information")

print("Path to dataset files:", path)
```

```python
import polars as pl

df_link = "/root/.cache/kagglehub/datasets/ekibee/car-sales-information/versions/14/all_regions.csv"

df = pl.read_csv(df_link,
		     encoding='utf8',
		     dtypes={'engineDisplacement': str})

df.head()
```

```python
df.select(['brand', 'color', 'mileage', 'price']).head(5)
```

```python
df.select([
    pl.col('price').max().alias('price_max'),
    pl.col('brand').n_unique().alias('brand_names_count'),
    pl.col('mileage').min().alias('mileage_min')
])
```

```python
df.filter((pl.col("brand") == "Toyota")
          & (pl.col("color") == "Черный")
          & (pl.col('year') > 2015) ).sort("price").limit(3)
```

```python
q = (
    df.lazy()
    .drop_nulls()
    .filter(
        (pl.col('brand') == 'Mercedes-Benz')
        & (pl.col('color') == 'Черный')
        & (pl.col('bodyType') == 'Седан')
        & (pl.col('year') > 2010))
    .sort(['year', 'price'])
    .limit(5)
)
%time q.collect()
```
