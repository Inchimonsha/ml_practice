```python
import numpy as np
import dask.array as da

# Создаем NumPy массив
numpy_array = np.arange(1000)

# Конвертируем в Dask Array с разбивкой на блоки по 100 элементов
dask_array = da.from_array(numpy_array, chunks=100)

# Выполняем вычисление суммы (отложенное до вызова compute)
result = dask_array.sum().compute()

print(result)
```
