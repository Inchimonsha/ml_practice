# Pros and Cons of CART

## Задание 1: Анализ преимуществ и недостатков CART

### Описание:
Напишите аналитический отчёт (1–2 страницы), в котором:

Перечислите и объясните преимущества CART (например, интерпретируемость, отсутствие необходимости нормализации данных).
Перечислите и объясните недостатки CART (например, склонность к переобучению, нестабильность).
Приведите примеры задач, в которых CART будет эффективен, и задач, в которых он не подходит.
Дополнительно:

Сравните CART с другими алгоритмами (например, логистическая регрессия, случайный лес).
