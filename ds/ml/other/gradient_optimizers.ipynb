```python
import numpy as np
import matplotlib.pyplot as plt

# Определение функции и ее производной
def func(x):
    return -0.5 * x + 0.2 * x ** 2 - 0.01 * x ** 3 - 0.3 * np.sin(4*x)

def dfunc(x):
    return -0.5 + 0.4 * x - 0.03 * x ** 2 - 1.2 * np.cos(4*x)

# Параметры градиентного спуска
learning_rate = 0.1
max_iterations = 1000
tolerance = 1e-6

# Инициализация начального значения x
x = -3.5

# Инициализация истории значений x и функции
x_history = [x]
func_history = [func(x)]

# Градиентный спуск
for _ in range(max_iterations):
    # Вычисление градиента
    gradient = dfunc(x)

    # Обновление x
    x_new = x - learning_rate * gradient

    # Проверка на достижение желаемой точности
    if np.abs(x_new - x) < tolerance:
        break

    # Обновление истории
    x = x_new
    x_history.append(x)
    func_history.append(func(x))

# Визуализация результатов
x_values = np.linspace(-5, 5, 400)
y_values = func(x_values)

plt.plot(x_values, y_values, label='Функция')
plt.plot(x_history, func_history, 'ro-', label='Градиентный спуск')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Градиентный спуск')
plt.legend()
plt.show()

print(f"Минимум функции достигнут в точке x = {x:.4f}")
```

```python
import numpy as np


def func(x):
    return 2 * x + 0.1 * x ** 3 + 2 * np.cos(3*x)

def dfunc(x):
    return 2 + 0.3 * x ** 2 - 6 * np.sin(3*x)

# здесь объявляйте функцию df (производную) и продолжайте программу
eda = 0.5
x = 4.0
xx = x
N = 200
alpha = 0.8
G = 0
eps = 0.01

for _ in range(N):
    xx = xx - eda * dfunc(x)
    G = alpha * G + (1 - alpha) * dfunc(x) * dfunc(x)
    x = x - eda * dfunc(x) / (np.sqrt(G) + eps)

print(x, xx)
```
