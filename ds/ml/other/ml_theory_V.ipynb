```python
# !pip install -r "source/requirements.yml"
import yaml
import subprocess

# Путь к вашему файлу YAML
yaml_file_path = '/content/drive/MyDrive/Colab_Notebooks/source/requirements.yml'

# Чтение файла YAML
with open(yaml_file_path, 'r') as file:
    dependencies = yaml.safe_load(file)

# Извлечение зависимостей
packages = dependencies['dependencies']

# Установка зависимостей с помощью pip
for package in packages:
    # Удаляем версию Python из списка пакетов
    if package.startswith('python'):
        continue
    subprocess.check_call(['pip', 'install', package])
!pip install lxml==4.5.1
```
