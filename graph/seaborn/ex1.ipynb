## Построение графиков с несколькими осями и масштабами (facetgrid, catplot)

Разбить данные на подгруппы по одному признаку и построить одинаковые графики по ним.

Пример: анализ продаж по разным регионам в отдельных панелях.

Цель: удобное сравнение групп в одном визуальном представлении.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Создаем синтетический набор данных с временными признаками
np.random.seed(42)
date_range = pd.date_range(start='2023-01-01', end='2025-12-31', freq='D')
n_days = len(date_range)

base_trend = np.linspace(100, 180, n_days)
seasonality = 20 * np.sin(np.pi * date_range.dayofyear / 183) + 40 * (date_range.month == 12)
noise = np.random.normal(0, 8, n_days)
weekly = 15 * (date_range.dayofweek < 5)

sales_data = base_trend + seasonality + noise + weekly

df_sales = pd.DataFrame({
    'date': date_range,
    'sales': sales_data
})
df_sales['year'] = df_sales['date'].dt.year
df_sales['month'] = df_sales['date'].dt.month

# Агрегируем по месяцам
monthly_sales = df_sales.groupby([df_sales['year'], df_sales['month']])['sales'].mean().reset_index()
monthly_sales['year_month'] = pd.to_datetime(monthly_sales['year'].astype(str) + '-' + monthly_sales['month'].astype(str))

# Построение lineplot с разделением по годам
plt.figure(figsize=(14, 8))
sns.lineplot(data=monthly_sales, x='month', y='sales', hue='year', palette='viridis', marker='o')
plt.xticks(ticks=range(1, 13), labels=['Янв','Фев','Мар','Апр','Май','Июн','Июл','Авг','Сен','Окт','Ноя','Дек'])
plt.title('Средние продажи по месяцам за разные года')
plt.xlabel('Месяц')
plt.ylabel('Средние продажи')
plt.show()
```
