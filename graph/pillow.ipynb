```python
from PIL import Image, ImageDraw, ImageFont

def create_schedule_card(schedule_text, filename="schedule_card.png"):
    width, height = 600, 400
    bg_color = (255, 255, 255)  # белый фон
    text_color = (0, 0, 0)      # черный текст

    img = Image.new('RGB', (width, height), color=bg_color)
    draw = ImageDraw.Draw(img)

    # Укажите путь к реальному ttf-файлу с поддержкой кириллицы
    # font_path = "arial.ttf"  # либо полный путь, например "C:/Windows/Fonts/arial.ttf"
    # font_size = 18
    # font = ImageFont.truetype(font_path, font_size)

    # Можно указать свой шрифт ttf, либо использовать дефолт
    font = ImageFont.load_default()

    # Пишем многострочный текст
    draw.multiline_text((20, 20), schedule_text, fill=text_color, font=font, spacing=4)

    img.save(filename)

# Пример вызова
schedule = """
📅 Понедельник, 1 сентября
1️⃣ 08:30-10:00
🏫 Алгоритмы и структуры данных
412, пр. Иванов

2️⃣ 10:15-11:45
🏢 Физика
108, доц. Петрова
"""
create_schedule_card(schedule)
```
