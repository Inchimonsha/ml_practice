## Проект «Исследование данных» с использованием публичных датасетов

Взять открытый набор данных (например, о погоде, COVID-19, экономике).

Выполнить полный анализ с визуализацией трендов, распределений, взаимосвязей.

Создать интерактивный отчет с веб-отображением графиков Plotly.

```python
import plotly.express as px
import pandas as pd

# Загружаем примерный датасет
df = px.data.gapminder().query("year == 2007")

# Строим интерактивный scatter plot ВВП на душу населения и продолжительности жизни
fig = px.scatter(df, x="gdpPercap", y="lifeExp", color="continent",
                 size="pop", hover_name="country",
                 log_x=True, size_max=60,
                 title="Жизненная продолжительность и ВВП по странам в 2007 году")

fig.show()
```
