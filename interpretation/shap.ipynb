```python
import shap
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor
from sklearn.datasets import fetch_california_housing

data = fetch_california_housing(as_frame=True)
X = data.data
y = data.target

model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X, y)

# Создание объекта-объяснителя SHAP (TreeExplainer для деревьев)
explainer = shap.TreeExplainer(model)

# Вычисление SHAP-значений для данных
shap_values = explainer.shap_values(X)

# Визуализация: summary_plot показывает важность признаков
shap.summary_plot(shap_values, X, plot_type="bar")  # бар-чарт общей важности

# Полный summary_plot с разбивкой по каждому признаку
shap.summary_plot(shap_values, X)
```

Такой пример показывает, какие признаки модели больше всего влияют на предсказания, и как именно. Визуализации помогают понять, например, где увеличение значения признака сказывается на повышении или понижении прогноза.

```python
# объяснение предсказания для конкретного объекта (водопад):
# Визуализация SHAP для отдельного примера (первого наблюдения)
shap.initjs()  # инициализация JavaScript для интерактивных графиков в Jupyter

shap.force_plot(explainer.expected_value, shap_values[0,:], X.iloc[0,:])
```

Здесь видно, как каждый признак влияет на конкретный прогноз модели.
